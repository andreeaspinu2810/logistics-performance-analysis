```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("../data/raw")

customers = pd.read_csv(DATA_DIR / "customers.csv")
routes = pd.read_csv(DATA_DIR / "routes.csv")
loads = pd.read_csv(DATA_DIR / "loads.csv")
trips = pd.read_csv(DATA_DIR / "trips.csv")
delivery_events = pd.read_csv(DATA_DIR / "delivery_events.csv")
```

```python
tables = {
    "customers" : customers,
    "routes" : routes,
    "loads" : loads,
    "trips" : trips,
    "delivery_events" : delivery_events}
```

```python
for name, df in tables.items(): 
    print(f"\n{name.upper()}")
    print("Shape:",df.shape)
    print("Columns:", df.columns.to_list)
    display(df.head(5))
```

missing values & duplicates

```python
for name, df in tables.items():
    print(f"\n{name.upper()}")
    print("Missing values:")
    print(df.isna().sum())
    print(f"\nDuplicate rows:", df.duplicated().sum())
```

data type

```python
for name, df in tables.items():
    print(f"\n{name.upper()}")
    print(df.dtypes)
```

PK, FK, CARDINALITIES

```python
for name, df in tables.items():
    print(f"{name}: {len(df):,} rows")
```

```python
primary_keys = {
    "customers": ("customer_id", customers),
    "routes": ("route_id", routes),
    "loads": ("load_id", loads),
    "trips": ("trip_id", trips),
    "delivery_events": ("event_id", delivery_events),
}

for table_name, (pk, df) in primary_keys.items():
    print(
        f"{table_name}: "
        f"{df[pk].nunique():,} unique {pk} / "
        f"{len(df):,} rows | "
        f"duplicates: {df[pk].duplicated().sum()}"
    )
```

```python
print(
    "loads → customers:",
    (~loads["customer_id"].isin(customers["customer_id"])).sum(),
    "orphan records"
)

print(
    "loads → routes:",
    (~loads["route_id"].isin(routes["route_id"])).sum(),
    "orphan records"
)

print(
    "trips → loads:",
    (~trips["load_id"].isin(loads["load_id"])).sum(),
    "orphan records"
)

print(
    "delivery_events → loads:",
    (~delivery_events["load_id"].isin(loads["load_id"])).sum(),
    "orphan records"
)

print(
    "delivery_events → trips:",
    (~delivery_events["trip_id"].isin(trips["trip_id"])).sum(),
    "orphan records"
)
```

```python
print("Loads per customer:")
print(loads.groupby("customer_id").size().describe())

print("\nTrips per load:")
print(trips.groupby("load_id").size().value_counts().sort_index())

print("\nDelivery events per load:")
print(delivery_events.groupby("load_id").size().value_counts().sort_index())
```

convert date columns from object to datetime

```python
loads["load_date"] = pd.to_datetime(loads["load_date"])

trips["dispatch_date"] = pd.to_datetime(trips["dispatch_date"])

delivery_events["scheduled_datetime"] = pd.to_datetime(
    delivery_events["scheduled_datetime"]
)

delivery_events["actual_datetime"] = pd.to_datetime(
    delivery_events["actual_datetime"]
)
```
